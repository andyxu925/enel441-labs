# What is frequency domain?

Based on two facts:
- every signal can be expressed as a weighted sum of sinusoids of different frequencies,
- for a sinusoidal input, the output of a LTI system is a sinusoid at the same frequency, but with different magnitude and phase.

# Bode Plot - Visualizing the Frequency Domain

Bode plot consists of two plots: 
- x-axis of both plots is frequency (usually in radians). Axis is logarithmic.
- y-axis of top plot is called the Magnitude plot.
- y-axis of bottom plot is called Phase plot.

To construct a Bode Plot of $G(s)$:
1. excite the system with a sinusoid of frequency $\omega$ with unit magnitude and known phase
2. determine the magnitude of phase shift of the output sinusoid
3. plot the magnitude of the output on the top plot (a point at frequency $\omega$)
4. plot the phase shift of the output on the bottom plot
5. Return to step one for a new frequency $\omega$.


## What is Laplace Domain?

Based on the concept of a transfer function. A transfer function is a way of encoding the governing differential equation of a system. A transfer function has poles and zeros. 

In the lectures we have learned that to represent a transfer function $G(s)$ in Frequency domain, all we have to do is set $s = j\omega$. In this lab we will fit a transfer function to the Bode Plot data that you will collect from the physical system.

# Detrending the Output

You may notice the the output signal is not zero mean, and it can have a slight drift (or trend). 

The reason why the signal is not zero mean is because we have assigned the starting position to be $0$ rad. 

The main reason for the drift is because the gears and the connections between the rods are not perfect and can wiggle a little bit, causing a drift over time. 

We want to remove this drift, and also center the signal so that it is zero-mean. The scipy package has a function that does this for us. It is called sp.detrend. 

Run the following cell to plot the detrended data.

```python
detrended_disk_position = sp.signal.detrend(disk_position_measured)


fig,ax = plt.subplots(1)
ax.plot(t_measured,disk_position_measured,label='Measured Disk Position')
ax.plot(t_measured,detrended_disk_position,label='Measured Disk Position - detrended')
ax.legend()
ax.set_title('Measured Output')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Position of disk (rad)')
```

# Determining the Magnitude of the Output

In this lab, we are interested in the magnitude of the output sinusoid. 

There are several ways of determining the magnitude of the output. We will look at two of them; a time domain method and a frequency domain method. This will further emphasize to you the connection between time domain and frequency domain.

# Time Domain Approach

From the detrended measured sinusoid, determine the maximum (or minimum) value of the signal. This is the magnitude of the sinusoid. 

Note: if your output had a significant transient response, make sure that you do not include this portion of the signal in your calculation of the amplitude of the sinusoid. We are only interested in the steady state behavior of the system.

```python
# Ignore the first 20% of the signal to avoid any initial transient response
start_index = int(0.2 * len(detrended_disk_position))

# Extract the steady-state portion
steady_state_output = detrended_disk_position[start_index:]

# Find the maximum and minimum values
y_max = np.max(steady_state_output)
y_min = np.min(steady_state_output)

# Estimate the magnitude (amplitude) of the output sinusoid
output_magnitude = (y_max - y_min) / 2

print("Maximum value =", y_max, "rad")
print("Minimum value =", y_min, "rad")
print("Magnitude of output sinusoid =", output_magnitude, "rad")
```

# Frequency Domain Approach

The time domain approach is sensitive to trends in the data. We can also determine the magnitude of the sinusoid from the frequency domain representation of the signal. 

In Python we can calculate the Fourier Transform using the np.fft.fft function. 

What should the Fourier Transform of a single sinusoid look like?

Use the code below to plot the Fourier Transform of the measured sinsoid and see if it matches your expectation.

Consider a few comments:
- The DC offset of the signal will appear at $f=0$ (a sinusoid with frequency of $0$ Hz).
- For the given sampling frequency of the system ($100$ samples/s the maximum frequency that can be represented is $50$ Hz (you will learn more about this in digital filtering or digital control)
- Frequency resolution of the sample points is $100$ samples/s / $1000$ samples  $= 0.1$ Hz.
- The second half of the plot is a mirror image of the first half, so you only should look at the first $500$ points. 

Note: calculate the Fourier Transform of the detrended measured output. If you use the undetrended output you will see a large DC gain, and low frequency smearing. Why?

```python
Y = np.fft.fft(detrended_disk_position)
f_vec = np.linspace(0,100-100/N,N)

fig,ax = plt.subplots(1)
ax.plot(f_vec[0:500],np.absolute(Y[0:500])/N*2)
ax.set_title('Fourier Transform of Measured Position')
ax.set_xlabel('Frequency (Hz)')
ax.set_ylabel('Magnitude (rad)')
```

# Constructing the Magnitude Portion of the Bode Plot

In the remainder of the lab you will repeat the procedure that you just did for a $1.5$ Hz sinusoid for sinusoids with different frequencies. 

Use the code above to apply sinusoids with the following frequencies to the motor: $0.3$ Hz to $8$ Hz in steps of $0.2$ Hz. 

Use the frequency domain method of calculating the magnitude of the sinusoid. It is more robust (you will get better estimates of the magnitude). 

Make sure that you always plot the measured signal and see what it looks like. 

You will need to write some code to store the magnitude, frequency, and the complex magnitude of the sinusoid for each experiment. The complex magnitude is just the value of the frequency response at the frequency of interest (the frequency that you excited the system at). It encodes both the magnitude and phase of the sinusoid. Let $a = b +jc$ be the complex magnitude of the sinusoid, then:
$$
\begin{align*}
Mag &= \text{abs}(a) = \sqrt{b^2+c^2} \\
Phase &= \text{angle}(a) = \text{atan}(\frac{c}{b})
\end{align*}
$$

Important note: it is very worthwhile to store each measured data set. This will help so that you can go back and fix a post-processing error after the fact, without having to redo the experiment. 

Once you have finished applying all the sinusoids to the system, you will plot the measured magnitude vs. frequency to construct the magnitude portion of a Bode plot. 

As you are applying the sinusoids keep the following questions in the back of your mind:
- What do you notice for the low frequencies?
- Do you notice anything different happening at around $2$ - $3$ Hz?
- What do you notice for the high frequencies ($4$ - $5$ Hz)?

```python
# ============================================================
# AUTOMATIC FREQUENCY SWEEP FOR BODE MAGNITUDE EXPERIMENT
# ============================================================

# Frequencies: 0.3 Hz up to 3.9 Hz in increments of 0.2 Hz
frequencies = np.arange(0.3, 3.9, 0.2)

# Input amplitude
input_amplitude = 100

# Storage arrays
measured_magnitudes = np.zeros(len(frequencies))
complex_frequency_response = np.zeros(len(frequencies), dtype=complex)

# Store all raw experimental data
all_measured_data = []

print("Number of experiments:", len(frequencies))
print("Frequencies:")
print(np.round(frequencies, 1))


# ============================================================
# LOOP THROUGH ALL FREQUENCIES
# ============================================================

for k, f in enumerate(frequencies):

    print("\n" + "=" * 60)
    print(
        f"Experiment {k + 1}/{len(frequencies)} "
        f"— Frequency = {f:.1f} Hz"
    )
    print("=" * 60)

    # --------------------------------------------------------
    # 1. Generate sinusoidal input
    # --------------------------------------------------------

    u1 = input_amplitude * np.sin(
        2 * np.pi * f * t
    )

    # Convert signed input into magnitude and direction
    u1_mag, u1_dir = lu.signal_to_mag_and_dir(u1)


    # --------------------------------------------------------
    # 2. Run the physical experiment
    # --------------------------------------------------------

    measured_outputs = lu.run_motor_and_disk_open_loop_streaming(
        u1_mag,
        u1_dir,
        com_port=arduino_com_port
    )


    # --------------------------------------------------------
    # 3. Convert encoder measurements into radians and time
    # --------------------------------------------------------

    disk_position_measured = (
        measured_outputs[0]
        * counts_per_revolution_to_rad
    )

    dt_measured = measured_outputs[1]

    # Convert accumulated time from ms to seconds
    t_measured = np.cumsum(dt_measured) / 1000.0


    # --------------------------------------------------------
    # 4. Detrend measured disk position
    # --------------------------------------------------------

    detrended_disk_position = sp.signal.detrend(
        disk_position_measured
    )


    # --------------------------------------------------------
    # 5. Remove initial transient
    # --------------------------------------------------------

    # Ignore first 20% of the experiment
    start_index = int(
        0.20 * len(detrended_disk_position)
    )

    t_ss = t_measured[start_index:]

    y_ss = detrended_disk_position[start_index:]


    # --------------------------------------------------------
    # 6. Construct the corresponding input at measured times
    # --------------------------------------------------------

    u_ss = input_amplitude * np.sin(
        2 * np.pi * f * t_ss
    )


    # --------------------------------------------------------
    # 7. Frequency-domain calculation
    # --------------------------------------------------------

    # Complex Fourier coefficient of output
    Y_complex = (
        2.0 / len(y_ss)
    ) * np.sum(
        y_ss
        * np.exp(
            -1j * 2 * np.pi * f * t_ss
        )
    )

    # Complex Fourier coefficient of input
    U_complex = (
        2.0 / len(u_ss)
    ) * np.sum(
        u_ss
        * np.exp(
            -1j * 2 * np.pi * f * t_ss
        )
    )

    # Experimental frequency response:
    #
    #       G(jw) = Y(jw) / U(jw)
    #
    G_complex = Y_complex / U_complex

    # Magnitude of frequency response
    G_magnitude = np.abs(G_complex)


    # --------------------------------------------------------
    # 8. Store results
    # --------------------------------------------------------

    measured_magnitudes[k] = G_magnitude

    complex_frequency_response[k] = G_complex

    all_measured_data.append(
        {
            "frequency": f,
            "time": t_measured.copy(),
            "input": u1.copy(),
            "disk_position": disk_position_measured.copy(),
            "detrended_disk_position":
                detrended_disk_position.copy(),
            "complex_response": G_complex,
            "magnitude": G_magnitude
        }
    )


    # --------------------------------------------------------
    # 9. Display information for this experiment
    # --------------------------------------------------------

    print(
        f"Complex response = "
        f"{G_complex}"
    )

    print(
        f"Magnitude = "
        f"{G_magnitude:.6f}"
    )

    print(
        f"Phase = "
        f"{np.angle(G_complex, deg=True):.2f} degrees"
    )


    # --------------------------------------------------------
    # 10. Plot measured data for this frequency
    # --------------------------------------------------------

    fig, ax = plt.subplots(
        2,
        1,
        figsize=(9, 6)
    )

    # Input
    ax[0].plot(
        t,
        u1
    )

    ax[0].set_title(
        f"Input Applied to Motor — {f:.1f} Hz"
    )

    ax[0].set_xlabel(
        "Time (s)"
    )

    ax[0].set_ylabel(
        "Input"
    )


    # Measured output
    ax[1].plot(
        t_measured,
        disk_position_measured,
        label="Measured Disk Position"
    )

    ax[1].plot(
        t_measured,
        detrended_disk_position,
        label="Detrended Disk Position"
    )

    ax[1].set_title(
        f"Measured Output — {f:.1f} Hz"
    )

    ax[1].set_xlabel(
        "Time (s)"
    )

    ax[1].set_ylabel(
        "Disk Position (rad)"
    )

    ax[1].legend()

    fig.tight_layout()

    plt.show()


print("\n====================================")
print("Frequency sweep completed.")
print("====================================")
```

# Estimate a Transfer Function

In this section you will fit a transfer function to your measured Bode Plot. Use the following code to estimate the transfer function. Print out the estimated transfer function. 

Make a plot with both the measured frequency response, and the frequency response of the estimated transfer function. How do they compare? 

To obtain the frequency response of a transfer function use the command >>ct.frequency_response(). Check the control systems package documentation for details. 


```python
# Number of poles and zeros of the estimated transfer function
na = 4
nb = 2

# Measured complex frequency response from our experiments
G_measured = complex_frequency_response

# Convert frequency from Hz to angular frequency (rad/s)
omega_vec = 2 * np.pi * frequencies

# Estimate the transfer-function coefficients
theta = lu.fit_transfer_function_to_frequency_response(
    [na, nb],
    G_measured,
    omega_vec
)

# Construct the estimated transfer function
G_est = ct.tf(
    np.flip(theta[0:nb, 0]),
    np.concatenate(([1], np.flip(theta[nb:nb+na, 0])))
)

# Display the estimated transfer function
print("Estimated Transfer Function:")
print(G_est)
```

## Plot measured vs. estimated frequency response

```python
# Obtain the frequency response of the estimated transfer function
response = ct.frequency_response(G_est, omega_vec)

# Extract magnitude of estimated transfer function
G_est_magnitude = np.asarray(response.magnitude).squeeze()

# Measured magnitude
G_measured_magnitude = np.abs(G_measured)

# Plot measured and estimated frequency responses
plt.figure(figsize=(9, 6))

plt.semilogx(
    frequencies,
    G_measured_magnitude,
    'o',
    label='Measured Frequency Response'
)

plt.semilogx(
    frequencies,
    G_est_magnitude,
    '-',
    label='Estimated Transfer Function'
)

plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude |G(jω)|')
plt.title('Measured vs. Estimated Frequency Response')

plt.grid(True, which='both')
plt.legend()

plt.show()
```

# Compare to Mathematical Model

In the lectures we have derived an expression for a transfer function that represents a Two-Disk system like the one in this lab. Use the code below to plot the Bode plot for this system. The plot will not match the Bode plot you constructed, because we do not know the moment of inertia, friction and spring constant for the lab setup. But the over all picture should be similar. Do you notice any similarities?

```python
import control as ct
import numpy as np

J1 = 10
J2 = 5
k = 2
b = 0.5

num_sys = np.array([1])
den_sys = np.array([J1*J2/k, (J1+J2)*b/k, (J1+J2+b**2/k), 2*b, 0])

P = ct.tf(num_sys,den_sys)
print(P)

# Plot over the same frequency range used in the experiment (0.3–3.7 Hz)
omega = 2 * np.pi * np.logspace(np.log10(0.3), np.log10(3.7), 500)

out = ct.bode_plot(P, omega=omega, Hz=True, dB=True, grid=True)
```

```python
out = ct.bode_plot(
    P,
    dB=True,
    Hz=True,
    grid=True
)

plt.show()
```

## Method 1 - Estimating the Physical Parameters

Can we use our experimentally estimated transfer function to estimate the physical parameters

$$
J_1,\quad J_2,\quad k,\quad \text{and}\quad b?
$$

The answer is **yes, in principle**.

The main idea is to compare the coefficients of our experimentally estimated transfer function with the coefficients of the mathematical model of the two-disk system.

In other words, we will use:

$$
\boxed{
\text{Experimental Data}
\rightarrow
\text{Estimated Transfer Function}
\rightarrow
\text{Coefficient Matching}
\rightarrow
\text{Physical Parameters}
}
$$

Let's work through this step-by-step.

### Step 1: Start With the Mathematical Model

Earlier, we used the following transfer function to represent the two-disk system:

$$
G(s)
=
\frac{1}{
\frac{J_1J_2}{k}s^4
+
\frac{(J_1+J_2)b}{k}s^3
+
\left(J_1+J_2+\frac{b^2}{k}\right)s^2
+
2bs
}.
$$

Here,

- $J_1$ is the moment of inertia of Disk 1,
- $J_2$ is the moment of inertia of Disk 2,
- $k$ is the spring constant, and
- $b$ is the damping/friction coefficient.

These are the physical parameters that we would like to estimate.

### Step 2: Write the Estimated Transfer Function in a General Form

Suppose the denominator of our experimentally estimated transfer function can be written as

$$
a_4s^4+a_3s^3+a_2s^2+a_1s.
$$

Therefore,

$$
G_{\mathrm{est}}(s)
=
\frac{1}{
a_4s^4+a_3s^3+a_2s^2+a_1s
}.
$$

The coefficients are:

- $a_4$: coefficient multiplying $s^4$,
- $a_3$: coefficient multiplying $s^3$,
- $a_2$: coefficient multiplying $s^2$,
- $a_1$: coefficient multiplying $s$.

Now compare this denominator with the denominator of our physical two-disk model.

### Step 3: Match the Coefficients

The denominator of the estimated transfer function is

$$
a_4s^4+a_3s^3+a_2s^2+a_1s.
$$

The denominator of the mathematical model is

$$
\frac{J_1J_2}{k}s^4
+
\frac{(J_1+J_2)b}{k}s^3
+
\left(J_1+J_2+\frac{b^2}{k}\right)s^2
+
2bs.
$$

If the estimated transfer function represents the same physical system, the coefficients multiplying the same powers of $s$ can be matched.

Therefore,

$$
\boxed{
a_4=\frac{J_1J_2}{k}
}
$$

$$
\boxed{
a_3=\frac{(J_1+J_2)b}{k}
}
$$

$$
\boxed{
a_2=J_1+J_2+\frac{b^2}{k}
}
$$

and

$$
\boxed{
a_1=2b.
}
$$

We now have four equations that can be used to estimate the four unknown physical parameters:

$$
J_1,\quad J_2,\quad k,\quad b.
$$

### Step 4: Estimate the Damping Coefficient $b$

Start with the simplest equation:

$$
a_1=2b.
$$

Solving for $b$ gives

$$
\boxed{
b=\frac{a_1}{2}
}.
$$

Therefore, once we know the coefficient of $s$ in the estimated denominator, we can immediately estimate $b$.

```python
# Extract the denominator coefficients from the
# experimentally estimated transfer function

den = np.squeeze(G_est.den)

print("Estimated denominator coefficients:")
print(den)

# For a fourth-order denominator:
#
# a4*s^4 + a3*s^3 + a2*s^2 + a1*s + a0

a4 = den[0]
a3 = den[1]
a2 = den[2]
a1 = den[3]

print("\na4 =", a4)
print("a3 =", a3)
print("a2 =", a2)
print("a1 =", a1)

# Estimate the damping coefficient b
b_est = a1 / 2

print("\nEstimated b =", b_est)
```

### Step 5: Find $J_1+J_2$

The next equations contain $J_1+J_2$ several times.

To make the calculation easier, define

$$
S=J_1+J_2.
$$

The coefficient $a_3$ is

$$
a_3=\frac{Sb}{k}.
$$

Solving for $k$ gives

$$
k=\frac{Sb}{a_3}.
$$

We also know that

$$
a_2=S+\frac{b^2}{k}.
$$

Substitute

$$
k=\frac{Sb}{a_3}
$$

into this equation:

$$
a_2
=
S+\frac{b^2}{Sb/a_3}.
$$

Simplifying gives

$$
a_2
=
S+\frac{ba_3}{S}.
$$

Multiplying both sides by $S$ gives

$$
a_2S=S^2+ba_3.
$$

Rearranging,

$$
\boxed{
S^2-a_2S+ba_3=0
}.
$$

This is a quadratic equation in $S$.

Using the quadratic formula,

$$
\boxed{
S=
\frac{a_2\pm\sqrt{a_2^2-4ba_3}}{2}
}.
$$

Recall that

$$
S=J_1+J_2.
$$

```python
# Calculate the discriminant of the quadratic equation for S
discriminant_S = a2**2 - 4*b_est*a3

# Calculate the two possible solutions
S_plus = (a2 + np.sqrt(discriminant_S)) / 2
S_minus = (a2 - np.sqrt(discriminant_S)) / 2

print("Possible values of J1 + J2:")
print("S_plus  =", S_plus)
print("S_minus =", S_minus)
```

### Step 6: Estimate the Spring Constant $k$

Once a physically meaningful value of

$$
S=J_1+J_2
$$

has been selected, we can use

$$
a_3=\frac{Sb}{k}.
$$

Solving for $k$ gives

$$
\boxed{
k=\frac{Sb}{a_3}
}.
$$

For this calculation, we will select the physically meaningful positive solution for $S$.

```python
# Select a physically meaningful positive value of S
#
# Check S_plus and S_minus above before making this choice.
S = S_plus

# Estimate the spring constant
k_est = S * b_est / a3

print("Estimated J1 + J2 =", S)
print("Estimated k =", k_est)
```

### Step 7: Find the Product $J_1J_2$

From the coefficient of $s^4$,

$$
a_4=\frac{J_1J_2}{k}.
$$

Therefore,

$$
\boxed{
J_1J_2=a_4k
}.
$$

At this point, we know both

$$
J_1+J_2=S
$$

and

$$
J_1J_2=a_4k.
$$

Knowing the sum and product allows us to determine $J_1$ and $J_2$ individually.

```python
# Calculate the product J1*J2
J_product = a4 * k_est

print("J1 + J2 =", S)
print("J1 * J2 =", J_product)
```

### Step 8: Estimate $J_1$ and $J_2$

We now know the sum

$$
J_1+J_2=S
$$

and the product

$$
J_1J_2=a_4k.
$$

If two numbers have a sum $S$ and product $P$, they are the roots of

$$
x^2-Sx+P=0.
$$

For our system,

$$
x^2-Sx+a_4k=0.
$$

Using the quadratic formula,

$$
\boxed{
J_{1,2}
=
\frac{S\pm\sqrt{S^2-4a_4k}}{2}
}.
$$

Therefore, the two solutions provide estimates of the moments of inertia $J_1$ and $J_2$.

```python
print("S =", S)
print("J_product =", J_product)

discriminant_J = S**2 - 4*J_product

print("Discriminant =", discriminant_J)
```

```python
# Calculate the discriminant
discriminant_J = S**2 - 4*J_product

print("Discriminant =", discriminant_J)

# Check whether real-valued physical parameters are possible
if discriminant_J >= 0:

    # Estimate J1 and J2
    J1_est = (S + np.sqrt(discriminant_J)) / 2
    J2_est = (S - np.sqrt(discriminant_J)) / 2

    print("Estimated J1 =", J1_est)
    print("Estimated J2 =", J2_est)

else:

    print("No real-valued J1 and J2 can be obtained.")
    print("The estimated transfer function does not exactly")
    print("satisfy the assumed physical two-disk model.")
```

### Putting Everything Together

We have now used the experimentally estimated transfer function to obtain estimates of the physical parameters of the two-disk system.

The process was

$$
\boxed{
G_{\mathrm{est}}(s)
\rightarrow
a_4,a_3,a_2,a_1
\rightarrow
b
\rightarrow
J_1+J_2
\rightarrow
k
\rightarrow
J_1J_2
\rightarrow
J_1,J_2
}
$$

Let's display all of the estimated parameters together.

```python
print("Estimated Physical Parameters")
print("--------------------------------")
print("J1 =", J1_est)
print("J2 =", J2_est)
print("k  =", k_est)
print("b  =", b_est)
```

### Important Note

This approach demonstrates how **system identification can connect measured input-output data to physical system parameters**.

However, experimental measurements contain noise and modelling uncertainties. In addition, the transfer function estimated earlier in this lab may be more general than the exact mathematical structure assumed for the two-disk model.

Therefore, the calculated values of

$$
J_1,\quad J_2,\quad k,\quad b
$$

should be interpreted as **estimates**, rather than exact measurements of the physical parameters.

If the estimated transfer function does not closely follow the assumed two-disk model structure, direct coefficient matching may also produce inaccurate or non-physical parameter values.

### Takeaway

The important idea is that a transfer function is not only a mathematical description of input-output behaviour. When we know the underlying physical model, its coefficients can also contain information about physical quantities such as

- moments of inertia,
- damping or friction, and
- spring stiffness.

Thus, experimental frequency-response measurements can potentially be used to learn something about the physical system itself.

### What if the Discriminant is Negative?

For physically meaningful moments of inertia, $J_1$ and $J_2$ must be
real and positive.

Recall that

$$
J_{1,2}
=
\frac{S\pm\sqrt{S^2-4J_1J_2}}{2}.
$$

Therefore, a real-valued solution requires

$$
\boxed{
S^2-4J_1J_2\geq0
}.
$$

If the calculated discriminant is negative, the estimated transfer
function cannot be exactly represented by the assumed physical
two-disk model using real-valued $J_1$ and $J_2$.

This can occur because the transfer function obtained from experimental
frequency-response data is an approximate model affected by measurement
noise, modelling error, and the particular transfer-function structure
used during system identification.

Thus, direct coefficient matching does not always produce physically
meaningful parameter estimates.

This is an important distinction:

$$
\boxed{
\text{A transfer function that fits experimental data well}
\not\Rightarrow
\text{its coefficients necessarily correspond directly to physical parameters.}
}
$$

## Going Further (Bonus) — Estimating Physical Parameters From the Measured Data

Earlier, we estimated a general transfer function from the measured frequency-response data.

Because we used

$$
n_a = 4,
\qquad
n_b = 2,
$$

the estimated model has the general form

$$
G_{\mathrm{est}}(s)
=
\frac{b_1s+b_0}
{s^4+a_3s^3+a_2s^2+a_1s+a_0}.
$$

This is a general input-output model obtained from the measured frequency-response data.

However, the mathematical two-disk model derived in the lectures has the more specific mechanical structure

$$
G_{\mathrm{mech}}(s)
=
\frac{1}{
\frac{J_1J_2}{k}s^4
+
\frac{(J_1+J_2)b}{k}s^3
+
\left(
J_1+J_2+\frac{b^2}{k}
\right)s^2
+
2bs
}.
$$

In the experiment, however, the applied input is a motor command rather than the idealized mechanical input used directly in the mathematical derivation.

Therefore, we introduce an additional gain $K$ to represent the relationship between the applied motor command and the mechanical excitation of the system.

The model used for fitting the measured data is therefore

$$
\boxed{
G_{\mathrm{physical}}(s)
=
\frac{K}{
\frac{J_1J_2}{k}s^4
+
\frac{(J_1+J_2)b}{k}s^3
+
\left(
J_1+J_2+\frac{b^2}{k}
\right)s^2
+
2bs
}
}
$$

where

- $J_1$ is the moment of inertia of Disk 1,
- $J_2$ is the moment of inertia of Disk 2,
- $k$ is the spring constant,
- $b$ is the damping coefficient, and
- $K$ is an effective input/actuator gain.

Notice that this model preserves the physical structure of the two-disk system while allowing for the scaling between the experimental motor command and the mechanical input.

Therefore, instead of directly matching the coefficients of the general experimentally fitted transfer function, we will estimate

$$
K,\quad J_1,\quad J_2,\quad k,\quad b
$$

by fitting this physical model directly to the measured complex frequency-response data.

This allows the estimated model to account for both the magnitude and phase of the experimentally measured response.

```python
# Display the general transfer function that was estimated earlier
print("Experimentally estimated general transfer function:")
print(G_est)

# Extract numerator and denominator coefficients
num_est = np.squeeze(G_est.num)
den_est = np.squeeze(G_est.den)

print("\nEstimated numerator coefficients:")
print(num_est)

print("\nEstimated denominator coefficients:")
print(den_est)
```

## Fit the Physical Model Directly

For every measured frequency $\omega$, the theoretical frequency response of the fitted physical model is

$$
G(j\omega)
=
\frac{K}{
\frac{J_1J_2}{k}(j\omega)^4
+
\frac{(J_1+J_2)b}{k}(j\omega)^3
+
\left(
J_1+J_2+\frac{b^2}{k}
\right)(j\omega)^2
+
2b(j\omega)
}.
$$

Here,

- $J_1$ is the moment of inertia of Disk 1,
- $J_2$ is the moment of inertia of Disk 2,
- $k$ is the spring constant,
- $b$ is the damping coefficient, and
- $K$ is an effective input/actuator gain that accounts for the scaling between the applied motor command and the mechanical input to the two-disk system.

We have already measured the complex frequency response

$$
G_{\mathrm{measured}}(j\omega).
$$

We can therefore search for positive values of

$$
K,\qquad J_1,\qquad J_2,\qquad k,\qquad b
$$

that make the physical model as close as possible to the measured frequency-response data.

Because the frequency response is complex, it contains both **magnitude** and **phase** information. Therefore, both parts of the measured response can be used when estimating the physical parameters.

The objective is therefore to find

$$
\boxed{
K,\;J_1,\;J_2,\;k,\;b
}
$$

such that

$$
G_{\mathrm{physical}}(j\omega)
\approx
G_{\mathrm{measured}}(j\omega)
$$

over the range of frequencies tested experimentally.

```python
from scipy.optimize import least_squares
import numpy as np
import matplotlib.pyplot as plt
import control as ct

# Convert experimental frequencies from Hz to rad/s
omega = 2 * np.pi * frequencies


def two_disk_response(params, omega):
    """
    Frequency response of the two-disk model.

    params = [K, J1, J2, k, b]

    K  = unknown input/actuator gain
    J1 = moment of inertia of disk 1
    J2 = moment of inertia of disk 2
    k  = spring constant
    b  = damping coefficient
    """

    K, J1, J2, k, b = params

    s = 1j * omega

    denominator = (
        (J1 * J2 / k) * s**4
        + ((J1 + J2) * b / k) * s**3
        + (J1 + J2 + b**2 / k) * s**2
        + 2 * b * s
    )

    return K / denominator
```

### Define the Fitting Error

For a perfect model, the physical-model frequency response would exactly match the measured frequency response:

$$
G_{\mathrm{physical}}(j\omega)
=
G_{\mathrm{measured}}(j\omega).
$$

In practice, experimental noise, modelling approximations, and other non-ideal effects mean that they will not match exactly.

Since a frequency response contains both **magnitude** and **phase** information, we compare the physical model with the measured data using both quantities.

For the magnitude, we calculate the logarithmic error

$$
e_{\mathrm{mag}}(\omega)
=
\ln\left|G_{\mathrm{physical}}(j\omega)\right|
-
\ln\left|G_{\mathrm{measured}}(j\omega)\right|.
$$

For the phase, we calculate the phase difference

$$
e_{\mathrm{phase}}(\omega)
=
\angle
\left(
\frac{G_{\mathrm{physical}}(j\omega)}
{G_{\mathrm{measured}}(j\omega)}
\right).
$$

The numerical optimization then searches for positive values of

$$
K,\qquad J_1,\qquad J_2,\qquad k,\qquad b
$$

that minimize these magnitude and phase errors over all of the experimentally tested frequencies.

In this way, the fitted physical model attempts to reproduce both the **size** and the **phase shift** of the measured system response.

```python
def fitting_error(log_params, omega, G_measured):
    """
    Error used for fitting magnitude and phase.

    Parameters are optimized in logarithmic form so that
    K, J1, J2, k and b always remain positive.
    """

    # Convert back from logarithmic parameters
    params = np.exp(log_params)

    G_model = two_disk_response(params, omega)

    # ----- Magnitude error -----
    magnitude_error = (
        np.log(np.abs(G_model))
        - np.log(np.abs(G_measured))
    )

    # ----- Phase error -----
    # angle(G_model / G_measured) gives the phase difference
    phase_error = np.angle(
        G_model / G_measured
    )

    # Combine magnitude and phase errors
    return np.concatenate(
        [magnitude_error, phase_error]
    )
```

```python
initial_guess = np.array([
    1.0,    # K
    10.0,   # J1
    5.0,    # J2
    2.0,    # k
    0.5     # b
])

# Optimize logarithms of the parameters
log_initial_guess = np.log(initial_guess)

result = least_squares(
    fitting_error,
    log_initial_guess,
    args=(omega, G_measured),
    max_nfev=50000
)

# Convert fitted logarithmic parameters back
K_est, J1_est, J2_est, k_est, b_est = np.exp(result.x)

print("Estimated Parameters")
print("--------------------------------")
print(f"K  = {K_est:.6f}")
print(f"J1 = {J1_est:.6f}")
print(f"J2 = {J2_est:.6f}")
print(f"k  = {k_est:.6f}")
print(f"b  = {b_est:.6f}")

print("\nOptimization successful:", result.success)
print("Optimization cost:", result.cost)
```

### Construct the Transfer Function Using the Estimated Physical Parameters

Now that we have estimates of

$$
K,\qquad J_1,\qquad J_2,\qquad k,\qquad b,
$$

we can substitute them back into the fitted physical two-disk model:

$$
G_{\mathrm{physical}}(s)
=
\frac{K}{
\frac{J_1J_2}{k}s^4
+
\frac{(J_1+J_2)b}{k}s^3
+
\left(
J_1+J_2+\frac{b^2}{k}
\right)s^2
+
2bs
}.
$$

Here, $K$ represents the effective input/actuator gain, while
$J_1$, $J_2$, $k$, and $b$ represent the physical parameters of the
two-disk system.

Substituting the estimated values into this expression gives the
estimated physically structured transfer function for the experimental system.

```python
# Numerator of the fitted physical model
# K_est represents the estimated effective input/actuator gain
num_physical = np.array([K_est])


# Denominator obtained from the estimated physical parameters
den_physical = np.array([
    J1_est * J2_est / k_est,
    (J1_est + J2_est) * b_est / k_est,
    J1_est + J2_est + b_est**2 / k_est,
    2 * b_est,
    0
])


# Construct the estimated physical two-disk transfer function
G_physical_est = ct.tf(
    num_physical,
    den_physical
)


# Display the resulting transfer function
print("Estimated Physical Two-Disk Transfer Function:")
print(G_physical_est)
```

```python
G_physical_response = two_disk_response(
    [K_est, J1_est, J2_est, k_est, b_est],
    omega
)

plt.figure(figsize=(9, 6))

plt.semilogx(
    frequencies,
    np.abs(G_measured),
    'o',
    label='Measured Frequency Response'
)

plt.semilogx(
    frequencies,
    np.abs(G_physical_response),
    '-',
    label='Fitted Two-Disk Model'
)

plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude |G(jω)|')
plt.title('Measured vs. Fitted Two-Disk Model')

plt.grid(True, which='both')
plt.legend()

plt.show()
```

```python
measured_phase = np.unwrap(
    np.angle(G_measured)
) * 180 / np.pi

model_phase = np.unwrap(
    np.angle(G_physical_response)
) * 180 / np.pi

plt.figure(figsize=(9, 6))

plt.semilogx(
    frequencies,
    measured_phase,
    'o',
    label='Measured'
)

plt.semilogx(
    frequencies,
    model_phase,
    '-',
    label='Fitted Two-Disk Model'
)

plt.xlabel('Frequency (Hz)')
plt.ylabel('Phase (degrees)')
plt.title('Measured vs. Fitted Two-Disk Model — Phase')

plt.grid(True, which='both')
plt.legend()

plt.show()
```

```python
num_physical = np.array([
    K_est
])

den_physical = np.array([
    J1_est * J2_est / k_est,
    (J1_est + J2_est) * b_est / k_est,
    J1_est + J2_est + b_est**2 / k_est,
    2 * b_est,
    0
])

G_physical_est = ct.tf(
    num_physical,
    den_physical
)

print("Estimated Two-Disk Transfer Function:")
print(G_physical_est)
```

```python
# Calculate the fitted physical-model response
G_physical_response = two_disk_response(
    [K_est, J1_est, J2_est, k_est, b_est],
    omega
)

# Calculate relative fitting error
relative_error = (
    np.linalg.norm(G_physical_response - G_measured)
    /
    np.linalg.norm(G_measured)
)

print(f"Relative fitting error = {relative_error:.4f}")
print(f"Relative fitting error = {100*relative_error:.2f}%")
```
